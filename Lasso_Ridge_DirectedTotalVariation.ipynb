```python
# nodes = number of nodes in the network
# observed_y = observed values of y (endogenous outputs of SEM)
# observed_x = observed values of x (exogenous inputs of SEM)
# gamma = regularization parameter. Learned from data via cross-validation 
```

#### Explanations on Lasso-regularized optimization on CVXPY.

1. We define the parameter we want to learn (here A).
2. We define the cost function (objective function).
3. We tell the compiler that we want to minimize the cost.
4. We define the constraints of the optimization: the diagonal of $\mathbf{A}$ is set to zero (no self-loops), and all entries of $\mathbf{A}$ are constrained to be non-negative.
5. We define the problem by combining the objective and the constraints.
6. We solve the optimization problem and store the optimal value of the cost function.
7. `A.value` contains the optimal estimate $\hat{\mathbf{A}}_{\mathrm{Lasso}}$ of the network matrix.

```python
# Lasso-regularized least squares optimization problem

A = cp.Variable((nodes,nodes)) # 1
cost     = cp.sum_squares(observed_y-A*observed_y-observed_x)+gamma*cp.atoms.norm1(A) # 2
obj      = cp.Minimize(cost) # 3
constr   = [cp.diag(A)==0,A[:,:]>=0] # 4
prob     = cp.Problem(obj,constr) # 5
opt_val  = prob.solve() # 6
solution_L1Norm = A.value # 7
```
